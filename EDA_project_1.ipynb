```python
# Cleaning
# EDA-4: Identify NaN values
import pandas as pd
# Identify null values
print(df.isnull())
# Count null values in each column
print(df.isnull().sum())
print("___________________________________________________________________________")
# Replace " ?" with NaN
df = df.replace(" ?", None)
# Check missing values after replacement
print(df.isna().sum())
print("___________________________________________________________________________")
# Fill missing values with a constant
print(df.fillna(0))
print("___________________________________________________________________________")
# Find the mean of numeric columns
print(df.mean(numeric_only=True))
print("___________________________________________________________________________")
# Find the median of numeric columns
print(df.median(numeric_only=True))
print("___________________________________________________________________________")
# Find the mode of each column
print(df.mode())
print("___________________________________________________________________________")
# Store the number of missing values
data = df.isna().sum()
print(data)
```

```python
# Identify duplicate rows
print(df.duplicated())
# Count the number of duplicate rows
print(df.duplicated().sum())
print("___________________________________________________________________________")
# Display only duplicate rows
print(df[df.duplicated()])
print("___________________________________________________________________________")
# Remove duplicate rows
df = df.drop_duplicates()
# Check again after removing duplicates
print(df.duplicated().sum())
# Display the cleaned DataFrame
print(df)
```

```python
#Calculate the first quartile (Q1), third quartile (Q3), and interquartile range (IQR) of Age.
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3-Q1
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
```

```python
# Identifying the Age values outside the IQR bounds and count them as potential outliers.
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3-Q1
lower = Q1 - 1.5 * IQR
UPPER = Q3 + 1.5 * IQR
iqr_outliers = df[
    (df['Age']< lower) |
    (df['Age']> UPPER)
]
print("IQR Outliers:", len(iqr_outliers))
```

```python
import pandas as pd
from scipy.stats import zscore
# Calculating the z-scores for Age and identify observations with an absolute z-score greater than 3 as outliers.
df['Age_zscore'] = zscore(df['Age'])
z_outliers = df[df['Age_zscore'].abs()>3]
print("z-score outliers:", len(z_outliers))
```

```python
#3. z_score trimming
df_trimmed = df[df['Age_zscore'].abs() <=3]
print(df_trimmed)
```

```python
#4.winsorization
from scipy.stats.mstats import winsorize
df['Age_winsorized'] = winsorize(
    df['Age'],
    limits=[0,0.3])
print(df.head())
```
